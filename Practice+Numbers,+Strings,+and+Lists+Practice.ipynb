You bought 500 shares of stock A at $\$$600 on day 0 and you sold it at  $\$$1000 on day 700.
The daily discount rate is 0.0001%. How much profit did you make in terms of net present value.


```python
shares = 500
buy_price = 600
sell_price = 1000
duration = 700
discount_rate_percent = 0.0001
discount_factor = 1/((1+(discount_rate_percent/100))**duration)
Net_present_profit = (sell_price*shares*discount_factor)- (buy_price*shares)
print(Net_present_profit)

```
